# **Questions:**

1) What is the significance of using data augmentations like resize, crop etc on training data?


ANS:
1. Improve Model Generalization
Augmentations simulate real-world variations (different sizes, lighting, positions, angles).

This helps the model learn more robust features that generalize better to unseen data.

2. Reduce Overfitting
By creating many different versions of the same image, the model is less likely to memorize specific training samples.

It acts as a form of regularization.

3. Handle Real-World Scenarios
Images in real-world applications are rarely perfect.

Resize, crop, flip, rotate, etc., simulate camera misalignments, occlusions, or different viewpoints.

4. Effectively Use Limited Data
When the dataset is small, augmentation helps create a larger effective training set without collecting new data.


| Augmentation                 | Purpose                                                       |
| ---------------------------- | ------------------------------------------------------------- |
| `Resize`                     | Normalize input size for consistent model input               |
| `RandomCrop`                 | Make the model learn parts of the object                      |
| `Flip (horizontal/vertical)` | Handle mirrored images or different orientations              |
| `Rotation`                   | Robustness to angular changes                                 |
| `ColorJitter`                | Adjust lighting, brightness, contrast                         |
| `Gaussian Noise`             | Help model handle noisy input                                 |
| `Normalization`              | Scale pixel values to a standard range (e.g., 0–1 or -1 to 1) |


2) What performance do you get if you don't use pretrained resnet model (Hint: Change pretrained=False and train the model)

Ans:
When training a ResNet (or any deep CNN) from scratch (pretrained=False), performance usually drops significantly unless you have a large and diverse dataset (like ImageNet).

1. Random Weights:
Without pretrained weights, the model starts with random initialization.

It has no knowledge of basic patterns (edges, textures, etc.).

2. Need for Massive Data:

Deep networks like ResNet have millions of parameters.

Training such models from scratch on small datasets leads to poor generalization.

3. Transfer Learning Helps:

Pretrained models have already learned to detect general features.

Fine-tuning them on a specific task leads to faster convergence and better accuracy.

3) If the resnet model was pre-trained on dataset significantly different than the ants vs bees data, would you still get good performance by using this pretrained model?

Ans:
1. Low-Level Features Are Generic:

Early layers of ResNet learn basic visual patterns: edges, corners, textures, colors.

These patterns are useful across almost all visual tasks—even if the objects differ (ants vs. cars, for example).

2. Transfer Learning Focuses on Adapting Later Layers:

You usually fine-tune only the last few layers of ResNet.

This allows the model to adapt high-level features specific to the new task (ants vs bees).

3. Speeds Up Training & Requires Less Data:

Even if not optimal, pretrained weights provide a head start compared to random initialization.

You can reach decent accuracy faster and with less data.
