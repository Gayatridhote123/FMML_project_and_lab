```python
import numpy as np
import pandas as pd
from numpy.linalg import svd
import matplotlib.pyplot as plt
```

```python
ratings = np.array([
    [5, 3, 0, 1],
    [4, 0, 0, 1],
    [1, 1, 0, 5],
    [1, 0, 0, 4],
    [0, 1, 5, 4],
])

print("Original Ratings Matrix:")
print(ratings)
```

```python

U, S, Vt = svd(ratings, full_matrices=False)
```

```python
print("\nU Matrix:\n", U)
print("\nSingular Values:\n", S)
print("\nV^T Matrix:\n", Vt)
```

```python
k = 2
S_k = np.diag(S[:k])
U_k = U[:, :k]
Vt_k = Vt[:k, :]
```

```python
ratings_approx = np.dot(np.dot(U_k, S_k), Vt_k)
print("\nReconstructed Matrix (k=2):")
print(np.round(ratings_approx, 2))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(ratings, cmap='viridis')
axes[0].set_title("Original Ratings")
```

```python
axes[1].imshow(ratings_approx, cmap='viridis')
axes[1].set_title("Reconstructed (k=2)")
plt.show()
```
